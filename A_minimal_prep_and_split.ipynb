```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

SEED = 42
```

```python
# Load Titanic dataset directly from URL
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# First look at the data
print("Shape:", df.shape)
print("\nFirst 5 rows:")
df.head()
```

```python
# Check what columns we have and their data types
print("Columns and types:")
print(df.dtypes)

print("\nMissing values per column:")
print(df.isnull().sum())
```

```python
# Drop columns that are identifiers or have too many missing values
# These carry no predictive signal for survival
cols_to_drop = ['PassengerId', 'Name', 'Ticket', 'Cabin']
df = df.drop(columns=cols_to_drop)

print("Remaining columns:", df.columns.tolist())
print("New shape:", df.shape)
```

```python
# Impute Age with median (more robust than mean for skewed data)
df['Age'] = df['Age'].fillna(df['Age'].median())

# Impute Embarked with mode (most frequent value)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Verify no missing values remain
print("Missing values after imputation:")
print(df.isnull().sum())
```

```python
# Convert Sex to binary: male = 0, female = 1
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# Convert Embarked to numbers: C = 0, Q = 1, S = 2
df['Embarked'] = df['Embarked'].map({'C': 0, 'Q': 1, 'S': 2})

print("Data types after encoding:")
print(df.dtypes)

print("\nFirst 5 rows after encoding:")
print(df.head())
```

```python
# Separate features (X) and target (y)
X = df.drop(columns=['Survived'])
y = df['Survived']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nSurvival rate:", round(y.mean() * 100, 2), "%")
```

```python
# Create stratified 80/20 train/test split
# Stratified means both train and test sets maintain the 38% survival rate
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,       # 20% for testing
    random_state=SEED,   # 42 — fixed for reproducibility
    stratify=y           # maintain survival ratio in both splits
)

print("Training set size:", X_train.shape)
print("Test set size:", X_test.shape)
print("\nTrain survival rate:", round(y_train.mean() * 100, 2), "%")
print("Test survival rate:", round(y_test.mean() * 100, 2), "%")
```

```python
# Save processed splits for use in Section B and C
# This ensures identical data across all notebooks
import numpy as np

np.save('X_train.npy', X_train.values)
np.save('X_test.npy', X_test.values)
np.save('y_train.npy', y_train.values)
np.save('y_test.npy', y_test.values)

print("Saved successfully:")
print("X_train.npy — shape:", X_train.shape)
print("X_test.npy — shape:", X_test.shape)
print("y_train.npy — shape:", y_train.shape)
print("y_test.npy — shape:", y_test.shape)
```

```python
# Final summary of Section A
print("=" * 50)
print("SECTION A — COMPLETE")
print("=" * 50)
print(f"\nOriginal dataset: 891 rows, 12 columns")
print(f"After cleaning: 891 rows, 7 features")
print(f"\nSteps completed:")
print(f"  ✅ Dropped: PassengerId, Name, Ticket, Cabin")
print(f"  ✅ Imputed: Age (median), Embarked (mode)")
print(f"  ✅ Encoded: Sex (0/1), Embarked (0/1/2)")
print(f"  ✅ Split: 80/20 stratified (seed=42)")
print(f"\nFinal splits:")
print(f"  X_train: {X_train.shape} | y_train: {y_train.shape}")
print(f"  X_test:  {X_test.shape} | y_test:  {y_test.shape}")
print(f"\nSurvival rates:")
print(f"  Original: 38.38%")
print(f"  Train:    {round(y_train.mean()*100,2)}%")
print(f"  Test:     {round(y_test.mean()*100,2)}%")
print(f"\nData saved: X_train, X_test, y_train, y_test (.npy)")
print("=" * 50)
```
